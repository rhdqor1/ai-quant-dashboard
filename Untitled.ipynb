```python
# 데이터 수집
import yfinance as yf
import pandas as pd

# 1. 수집할 기간 설정
start_date = "2020-01-01"
end_date = "2026-06-07"

print("🇺🇸미국, 🇰🇷한국, 🇨🇳중국 대표 ETF 데이터 수집을 시작합니다. 잠시만 기다려주세요...")

# 2. 미국 대표 기술주 ETF (QQQ)
us_etf = yf.Ticker("QQQ").history(start=start_date, end=end_date)
us_etf.rename(columns={'Close': 'US_QQQ_Close'}, inplace=True)
us_etf.index = us_etf.index.tz_localize(None)

# 3. 한국 대표 대형주 ETF (069500.KS: KODEX 200)
kr_etf = yf.Ticker("069500.KS").history(start=start_date, end=end_date)
kr_etf.rename(columns={'Close': 'KR_KODEX_Close'}, inplace=True)
kr_etf.index = kr_etf.index.tz_localize(None)

# 4. 중국 대표 대형주 ETF (FXI: iShares China Large-Cap ETF)
cn_etf = yf.Ticker("FXI").history(start=start_date, end=end_date)
cn_etf.rename(columns={'Close': 'CN_FXI_Close'}, inplace=True)
cn_etf.index = cn_etf.index.tz_localize(None)

# 5. 데이터 병합 (Outer Join) - 3국 데이터 합치기
merged_data = us_etf[['US_QQQ_Close']].join(kr_etf[['KR_KODEX_Close']], how='outer')
merged_data = merged_data.join(cn_etf[['CN_FXI_Close']], how='outer')

# 6. 결측치 정제 (각 나라마다 휴일이 다르기 때문에 앞의 데이터로 채워줌)
merged_data.ffill(inplace=True)
merged_data.dropna(inplace=True)

# 7. 결과를 CSV 파일로 저장
merged_data.to_csv("global_top3_etf_data.csv")

print("글로벌 3국 대표 ETF 데이터 병합 및 CSV 저장 완료! (파일명: global_top3_etf_data.csv)")
merged_data.tail() # 마지막 5줄 엿보기
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

print("데이터 전처리 및 방향성(UP/DOWN) 라벨링을 시작합니다")

# 1. 데이터 불러오기 (미국, 한국, 중국 데이터가 합쳐진 df)
data = df.values

# 2. 스케일링 (0과 1 사이로 압축)
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data)

# 3. 슬라이딩 윈도우 및 정답지(라벨) 생성
window_size = 10

X = [] # 문제지 (과거 10일치 데이터들)
y = [] # 정답지 (11일차의 '상승 여부')

# 데이터 전체를 돌면서 10일 단위로 묶어주기
for i in range(window_size, len(scaled_data)):
    X.append(scaled_data[i-window_size : i])
    
    # 핵심 수정 포인트: 주가를 그대로 넣지 않고, 전날보다 올랐는지 확인!
    # i번째 날(내일)의 미국 나스닥 가격이 i-1번째 날(오늘)보다 크면 1, 아니면 0
    if scaled_data[i, 0] > scaled_data[i-1, 0]:
        y.append(1)  # 상승
    else:
        y.append(0)  # 하락

X, y = np.array(X), np.array(y)

# 4. 훈련용(Train)과 테스트용(Test) 데이터 나누기
split_index = int(len(X) * 0.8)

X_train, X_test = X[:split_index], X[split_index:]
y_train, y_test = y[:split_index], y[split_index:]

print(f"전처리 완료!")
print(f"훈련용 문제지 크기: {X_train.shape}")
print(f"테스트용 정답지 예시 (1=상승, 0=하락): {y_train[:10]}")
```

```python
# 1. 딥러닝 라이브러리 설치 (이미 있다면 빠르게 넘어갑니다)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import matplotlib.pyplot as plt

print("글로벌 증시 방향성 예측 AI(LSTM) 구축을 시작합니다...")

# 2. 모델 설계도 만들기 (순차적으로 층을 쌓음)
model = Sequential()

# 3. 첫 번째 LSTM 층 (기억력을 담당하는 뇌세포 50개 배치)
# 입력 데이터 모양: (10일치 기간, 변수 개수)
model.add(LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2)) # 과적합(암기) 방지를 위해 20%의 뇌세포는 무작위로 쉬게 함

# 4. 두 번째 LSTM 층 (더 깊은 패턴을 찾기 위해 한 층 더 쌓음)
model.add(LSTM(units=50, return_sequences=False))
model.add(Dropout(0.2))

# 5. 최종 예측 층 (분류 모델로 핵심 변경!)
# '내일 오를 확률(0~1)'을 계산해야 하므로 활성화 함수로 'sigmoid'를 반드시 추가합니다.
model.add(Dense(units=1, activation='sigmoid'))

# 6. 모델 컴파일 (공부 방법 설정 - 분류용으로 변경)
# loss='binary_crossentropy': 이진 분류(상승/하락)에서 오차를 계산하는 전용 방식
# metrics=['accuracy']: 우리가 보고 싶어 하는 '59%의 정확도'를 화면에 띄워줍니다.
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("분류 모델 구조 완성! 본격적으로 과거 데이터를 보며 공부를 시작합니다.")

# 7. 모델 학습시키기 (fit)
history = model.fit(X_train, y_train, 
                    epochs=20, 
                    batch_size=32, 
                    validation_data=(X_test, y_test))

print("학습 완료!")

# 8. 학습 과정 시각화 (오차 대신 '정확도'가 올라가는 것을 시각화)
# 윈도우 환경 한글 폰트 깨짐 방지 세팅
# 8. 학습 과정 시각화 (Y축 고정해서 예쁘게 그리기)
plt.figure(figsize=(10, 5))

# 훈련 정확도와 검증(테스트) 정확도 그리기
plt.plot(history.history['accuracy'], label='훈련 데이터 정확도 (학습)', color='blue', linewidth=2)
plt.plot(history.history['val_accuracy'], label='실전 테스트 정확도 (검증)', color='red', linestyle='--', linewidth=2)

# [핵심] Y축의 범위를 0.45 ~ 0.65로 고정! (착시 현상 제거)
plt.ylim(0.45, 0.65) 

plt.title('AI 모델의 방향성 예측 정확도 (Accuracy)', fontsize=15)
plt.xlabel('반복 학습 횟수 (Epochs)')
plt.ylabel('정확도 (50% 기준치)')
plt.axhline(y=0.5, color='gray', linestyle=':', label='50% (동전 던지기)') # 50% 기준선 추가
plt.xticks(range(0, 21, 2))
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("AI 모델 실전 테스트 결과를 시각화합니다 (전문가용 디자인)...")

# 1. 예측 및 타율 계산
y_pred_prob = model.predict(X_test)
y_pred_binary = (y_pred_prob > 0.5).astype(int)
hit_ratio = np.mean(y_pred_binary[:, 0] == y_test)

# 2. 폰트 세팅
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False 

# 3. 도화지 준비 
plt.figure(figsize=(14, 6))
prob_1d = y_pred_prob.flatten()

# [디자인 핵심 1] Y축을 바짝 줌인해서 미세한 파동을 거대하게 보이게 만듭니다.
# 0.48부터 AI의 최대 확신도까지만 딱 잘라서 보여줍니다.
plt.ylim(0.48, max(prob_1d) + 0.01) 

# 50% 기준선 (강렬한 빨간색)
plt.axhline(y=0.5, color='crimson', linestyle='--', label='기본 매수 기준점 (50%)', linewidth=2)

# [디자인 핵심 2] AI의 상위 25% 확신도를 '강력 매수' 구간으로 분리
strong_buy_line = np.percentile(prob_1d, 75)
plt.axhline(y=strong_buy_line, color='goldenrod', linestyle=':', linewidth=2, label='강력 매수 기준선')

# 4. 확률 선 그리기 (두껍고 진한 네이비)
plt.plot(prob_1d, color='#1A5F7A', label='AI 예측 상승 확률', linewidth=2.5)

# [디자인 핵심 3] 확신도에 따라 형광펜(배경색)을 투톤으로 칠하기
# 일반 매수 구간 (50% ~ 강력매수선)
plt.fill_between(range(len(prob_1d)), 0.5, prob_1d, 
                 where=(prob_1d > 0.5), 
                 color='#22A699', alpha=0.25, label='일반 매수(Long) 유지 구간')

#  강력 매수 구간 (강력매수선 이상은 황금색으로 아주 진하게 칠함)
plt.fill_between(range(len(prob_1d)), strong_buy_line, prob_1d, 
                 where=(prob_1d >= strong_buy_line), 
                 color='#F2BE22', alpha=0.6, label='강력 매수 (Strong Buy) 구간')

# 5. 꾸미기
plt.title(f'한·미·중 동조화 기반 AI 퀀트 대시보드(실전 적중률: {hit_ratio * 100:.1f}%)', fontsize=18, fontweight='bold', pad=15)
plt.xlabel('시간 흐름 (테스트 기간, 일차)', fontsize=12)
plt.ylabel('AI 상승 확신도 (Probability)', fontsize=12)

# 범례(Legend) 디자인 깔끔하게 정리
plt.legend(loc='lower right', frameon=True, shadow=True, fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
